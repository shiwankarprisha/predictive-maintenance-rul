# 01. Exploratory Data Analysis (EDA)
## NASA C-MAPSS Turbofan Engine Degradation (FD001)
This notebook explores fleet lifespans, operational settings, and 21 multi-channel sensor distributions.

```
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_raw_train_data, get_dataset_summary
from src.rul import compute_run_to_failure_rul
```

```
raw_df = load_raw_train_data()
print(get_dataset_summary(raw_df))
raw_df.head()
```

```
# Lifespan distribution to failure
max_cycles = raw_df.groupby('engine_id')['cycle'].max()
plt.figure(figsize=(8, 4))
sns.histplot(max_cycles, bins=15, kde=True, color='royalblue')
plt.title('Turbofan Operating Lifespan to Failure')
plt.xlabel('Cycles to Failure')
plt.show()
```

```
# Correlation heatmap among sensors
sensor_cols = [f'sensor_{i}' for i in range(1, 22)]
plt.figure(figsize=(10, 8))
sns.heatmap(raw_df[sensor_cols].corr(), cmap='coolwarm', center=0)
plt.title('Raw Sensor Correlation Matrix')
plt.show()
```
